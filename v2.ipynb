```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [7200, 8400, 6900, 10300, 9100, 11800, 9700]

plt.figure(figsize=(8, 4))
plt.plot(
    days, steps,
    color="green",
    linewidth=2,
    marker="o",
    markerfacecolor="red",
    markeredgecolor="red"
)
plt.title("Daily Steps")
plt.xlabel("Day")
plt.ylabel("Steps")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Apr", "May", "Jun", "Jul", "Aug", "Sep"]
active_users = [120, 135, 142, 158, 171, 185]  # sample values, in thousands

plt.figure(figsize=(8, 4))
plt.plot(
    months, active_users,
    color="blue",
    linewidth=2,
    marker="o",
    markerfacecolor="red",
    markeredgecolor="red"
)
plt.title("Monthly Active Users — Music App")
plt.xlabel("Month")
plt.ylabel("Active users (thousands)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["May", "Jun", "Jul", "Aug", "Sep"]
zomato_orders = [82, 88, 93, 101, 109]
swiggy_orders = [76, 81, 90, 95, 104]  # sample values, in thousands

plt.figure(figsize=(8, 4))
plt.plot(months, zomato_orders, color="red", linestyle="-",
         marker="o", linewidth=2, label="Zomato")
plt.plot(months, swiggy_orders, color="orange", linestyle="--",
         marker="s", linewidth=2, label="Swiggy")

plt.title("Monthly Orders: Zomato and Swiggy")
plt.xlabel("Month")
plt.ylabel("Orders (thousands)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Apr", "May", "Jun", "Jul", "Aug", "Sep"]
data_gb = [18, 19.5, 20, 23, 22, 25]  # sample monthly usage

plt.figure(figsize=(8, 4))
plt.plot(months, data_gb, marker="o", linewidth=2)
plt.title("Monthly Data Usage Trend")
plt.xlabel("Month")
plt.ylabel("Data Used (GB)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
