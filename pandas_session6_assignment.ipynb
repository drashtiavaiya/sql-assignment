```python
import pandas as pd
import numpy as np

spotify_data = pd.DataFrame({
    "song": [
        "luther", "Good Days", "Pink + White",
        "luther", "As It Was", "Good Days"
    ],
    "artist": [
        "Kendrick Lamar & SZA", "SZA", "Frank Ocean",
        "Kendrick Lamar & SZA", "Harry Styles", "SZA"
    ],
    "streams": [2500000, 1800000, 950000, 2500000, 3200000, 1800000],
    "date": [
        "2026-09-01", "2026-09-02", "2026-09-03",
        "2026-09-01", "2026-09-04", "2026-09-02"
    ]
})

spotify_data.to_csv("spotify_streams.csv", index=False)

spotify = pd.read_csv("spotify_streams.csv")

cleaned_spotify = spotify.drop_duplicates(
    subset=["song"],
    keep="first"
)

print(cleaned_spotify)
```

```python
zomato_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "order_amount": [250, 320, 180, 400, 290, 350, 220, 2500]
})

Q1 = zomato_orders["order_amount"].quantile(0.25)
Q3 = zomato_orders["order_amount"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = zomato_orders[
    (zomato_orders["order_amount"] < lower_limit) |
    (zomato_orders["order_amount"] > upper_limit)
]

print("Q1:", Q1, "Q3:", Q3, "IQR:", IQR)
print("\nOutlier orders:")
print(outliers)
```

```python
from scipy.stats import zscore

ipl_tickets = pd.DataFrame({
    "date": pd.date_range("2026-04-01", periods=10),
    "tickets_sold": [950, 1020, 980, 1100, 1050, 990, 1010, 970, 1080, 5000]
})

ipl_tickets["z_score"] = zscore(ipl_tickets["tickets_sold"])

unusual_days = ipl_tickets[
    ipl_tickets["z_score"].abs() > 2
]

print(unusual_days)
```

```python
paytm_wallet = pd.DataFrame({
    "transaction_id": range(1, 11),
    "amount": [120, 250, 180, 300, 220, 150, 275, 200, 240, 9000]
})

def find_outlier_amounts(data, column):
    scores = zscore(data[column])
    return data.loc[abs(scores) > 2, column]

outlier_amounts = find_outlier_amounts(paytm_wallet, "amount")

print("Outlier transaction amounts:")
print(outlier_amounts)
```
