```python
import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
percentages = [28, 25, 18, 12, 17]

plt.figure(figsize=(7, 6))
plt.pie(
    percentages,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Share of Daily App Use")
plt.axis("equal")
plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [5000, 3200, 1800, 2500]  # sample monthly amounts in ₹

highest_index = spending.index(max(spending))
explode = [0.12 if i == highest_index else 0 for i in range(len(spending))]

plt.figure(figsize=(7, 6))
plt.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    explode=explode,
    startangle=90
)
plt.title("Monthly Spending by Category")
plt.axis("equal")
plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [5000, 3200, 1800, 2500]

fig, ax = plt.subplots(figsize=(7, 6))
ax.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    pctdistance=0.78,
    startangle=90
)

centre_circle = plt.Circle((0, 0), 0.55, color="white")
ax.add_artist(centre_circle)

ax.set_title("Monthly Spending by Category")
ax.axis("equal")
plt.show()
```

```python
import matplotlib.pyplot as plt

teams = ["CSK", "RCB", "MI", "KKR"]
followers_millions = [16, 14, 12, 8]

fig, ax = plt.subplots(figsize=(7, 6))
ax.pie(
    followers_millions,
    labels=teams,
    autopct="%1.1f%%",
    pctdistance=0.78,
    startangle=90,
    colors=["#F5C542", "#C62828", "#2874A6", "#6A4C93"]
)

centre_circle = plt.Circle((0, 0), 0.55, color="white")
ax.add_artist(centre_circle)

ax.set_title("IPL Teams Instagram Followers Share")
ax.axis("equal")
plt.show()
```
