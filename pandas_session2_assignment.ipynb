```python
import pandas as pd

ipl_url = "https://raw.githubusercontent.com/kkamal11/EDA_on_IPL/main/IPL_Matches_2008_2022.csv"

ipl_matches = pd.read_csv(ipl_url)

print(ipl_matches.head(10))
```

```python
mobile_expenses = pd.DataFrame({
    "Date": ["01-09-2026", "03-09-2026", "05-09-2026",
             "08-09-2026", "10-09-2026", "12-09-2026"],
    "Expense": ["Recharge", "Food", "Shopping",
                "Transport", "Internet", "Stationery"],
    "Amount": [299, 450, 1200, 180, 599, 250]
})

mobile_expenses.to_excel("mobile_expenses.xlsx", index=False)

expenses = pd.read_excel("mobile_expenses.xlsx")

print(expenses.tail(3))
```

```python
zomato_sample = pd.DataFrame({
    "Restaurant": [
        "Spice Villa", "Pizza Corner", "Green Leaf", "Urban Tadka",
        "Food Junction", "Royal Kitchen", "Cafe Aroma", "Tasty Bites"
    ],
    "City": [
        "Surat", "Surat", "Ahmedabad", "Vadodara",
        "Surat", "Rajkot", "Ahmedabad", "Surat"
    ],
    "Rating": [4.2, 4.0, 4.5, 3.8, 4.1, None, 4.3, 3.9],
    "Votes": [520, 340, 780, 210, 450, 190, 610, 275],
    "Average_Cost": [600, 450, 700, 500, 550, 800, 650, 400]
})

zomato_sample.to_csv("zomato_restaurants.csv", index=False)

zomato = pd.read_csv("zomato_restaurants.csv")

print("Dataset Information:")
zomato.info()

print("\nStatistical Summary:")
display(zomato.describe())

print("\nTwo Insights:")
print("1. Number of restaurants:", zomato["Restaurant"].nunique())
print("2. Average restaurant rating:", round(zomato["Rating"].mean(), 2))
```
