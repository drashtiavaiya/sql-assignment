```python
followers = pd.Series(
    [1200, 54000, 780, 150000, 32000],
    index=[
        "Aarav",
        "Diya",
        "Kabir",
        "Meera",
        "Rohan"
    ]
)

print("Follower counts:")
print(followers)

print(
    "\nHighest followers:",
    followers.idxmax(),
    "-",
    followers.max()
)

print(
    "Lowest followers:",
    followers.idxmin(),
    "-",
    followers.min()
)
```
