```python
product_name = "Redmi Note 12 Pro"

print("Uppercase:", product_name.upper())
print("Lowercase:", product_name.lower())
```

```python
def clean_brand_name(name):
    return name.strip().replace("-", " ")

brand_name = "  oneplus-Nord  "
print(clean_brand_name(brand_name))
```

```python
product = "Apple iPhone 14 Pro Max"

split_position = product.find(" ")
brand = product[:split_position]
model = product[split_position + 1:]

print("Brand:", brand)
print("Model:", model)
```

```python
def format_product_display(name, price):
    return f"{name} - ₹{price}"

formatted_product = format_product_display("Boat Earbuds", 1299)
print(formatted_product)
```

```python
messy_products = [
    " mi-Band 5 ",
    "SAMSUNG-Galaxy",
    "realme-Book "
]

cleaned_products = [
    product.strip().replace("-", " ").title()
    for product in messy_products
]

print("Cleaned Products:", cleaned_products)
```
