```python
import pandas as pd

instagram_users = pd.DataFrame({
    "username": ["CoolDude", "foodieQueen", "TravelWithMe", "ARTlover", "MusicFan"]
})

instagram_users["username"] = instagram_users["username"].str.lower()

print(instagram_users)
```

```python
song_titles = pd.Series([
    "Good Days Remix",
    "Nights",
    "Snooze Remix",
    "Pink + White",
    "Remix of luther Remix"
])

cleaned_titles = song_titles.str.replace(
    "Remix", "", regex=False
).str.strip()

print(cleaned_titles)
```

```python
zomato_reviews = pd.DataFrame({
    "review": [
        "The delivery was quick.",
        "Food tasted great.",
        "DELIVERY took too long.",
        "The packaging was neat.",
        "Excellent Delivery service."
    ]
})

delivery_reviews = zomato_reviews[
    zomato_reviews["review"].str.contains(
        "delivery", case=False, na=False
    )
]

print(delivery_reviews)
```

```python
categories = pd.Series([
    "Electronics/Mobiles/Smartphones",
    "Home/Kitchen/Appliances",
    "Fashion/Women/Dresses",
    "Electronics/Laptops/Gaming"
])

main_categories = categories.str.split("/").str[0]

print(main_categories)
```

```python
youtube_videos = pd.DataFrame({
    "title": [
        "Pandas basics for beginners",
        "How to learn SQL",
        "My data analytics journey",
        "Python project ideas"
    ]
})

youtube_videos["title"] = youtube_videos["title"].str.upper().str.replace(
    " ", "_", regex=False
)

print(youtube_videos)
```
