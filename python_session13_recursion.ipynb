```python
def reverse_string(text):
    if len(text) <= 1:
        return text

    return reverse_string(text[1:]) + text[0]


print(reverse_string("hello"))
```

```python
def sum_playlist(durations):
    if len(durations) == 0:
        return 0

    return durations[0] + sum_playlist(durations[1:])


song_durations = [210, 185, 240, 195]

print("Total playlist duration:", sum_playlist(song_durations), "seconds")
```

```python
count = 10  # Global variable


def update_count():
    count = 5  # Local variable
    print("Inside function:", count)


update_count()

print("Outside function:", count)
print("Explanation: The function prints the local value 5, while the global value remains 10.")
```

```python
def count_likes(posts):
    total_likes = 0

    for post in posts:
        total_likes += post["likes"]

        if "replies" in post:
            total_likes += count_likes(post["replies"])

    return total_likes


instagram_posts = [
    {
        "likes": 120,
        "replies": [
            {"likes": 25},
            {
                "likes": 30,
                "replies": [
                    {"likes": 10}
                ]
            }
        ]
    },
    {
        "likes": 200,
        "replies": [
            {"likes": 40}
        ]
    }
]

print("Total likes:", count_likes(instagram_posts))
```

```python
user_status = "Offline"  # Global variable

print("Before function:", user_status)


def show_whatsapp_status():
    user_status = "Online"  # Local variable
    print("During function:", user_status)


show_whatsapp_status()

print("After function:", user_status)
```
