```python
import re

chat_text = """
Call me at 9876543210.
The phone price is 45000.
Another seller uses 8123456789.
Order number is 123456.
You can also contact 7012345678.
Invalid number: 6123456789.
"""

phone_numbers = re.findall(r"\b[789]\d{9}\b", chat_text)

print("Valid Indian phone numbers:", phone_numbers)
```

```python
import re

def contains_valid_date(text):
    pattern = r"\b\d{2}/\d{2}/\d{4}\b"

    if re.search(pattern, text):
        return True
    else:
        return False


print(contains_valid_date("The delivery date is 25/06/2024."))
print(contains_valid_date("The delivery date is tomorrow."))
```

```python
import re

zomato_review = """
The food was good. Contact support at help@zomato.com.
My email is drashti22@gmail.com.
The restaurant replied from manager.restaurant@yahoo.in.
Invalid emails include user@ and @example.com.
"""

email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"

email_addresses = re.findall(email_pattern, zomato_review)

print("Valid email addresses:", email_addresses)
```

```python
import re

phone_number = "9876543210"

masked_number = re.sub(
    r"\d(?=\d{4})",
    "*",
    phone_number
)

print("Masked phone number:", masked_number)
```

```python
import re

order_pattern = r"\bOD\d{18}\b"

sample_orders = [
    "Your Flipkart order ID is OD123456789012345000.",
    "Track order OD987654321098765432 now.",
    "Invalid order ID: OD12345."
]

for order in sample_orders:
    match = re.search(order_pattern, order)

    if match:
        print("Valid order ID:", match.group())
    else:
        print("No valid order ID found.")
```
