```python
class Product:
    def __init__(self, price):
        self._price = price

    def get_price(self):
        return self._price

    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative.")

        self._price = new_price


product = Product(1499)

product.set_price(1299)
print("Updated price: ₹", product.get_price())

try:
    product.set_price(-500)
except ValueError as error:
    print("Error:", error)
```

```python
class Playlist:
    def __init__(self):
        self._songs = []

    def add_song(self, song):
        self._songs.append(song)

    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)

    def get_songs(self):
        return self._songs.copy()


playlist = Playlist()

playlist.add_song("Good Days")
playlist.add_song("Pink + White")
playlist.add_song("Luther")

playlist.remove_song("Pink + White")

print("Current playlist:", playlist.get_songs())
```

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"₹{amount} paid successfully using UPI.")


class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"₹{amount} paid successfully using Credit Card.")


upi_payment = UPI()
card_payment = CreditCard()

upi_payment.pay(750)
card_payment.pay(1200)
```
