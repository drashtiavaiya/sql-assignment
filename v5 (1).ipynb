```python
import matplotlib.pyplot as plt

runs = [142, 176, 198, 163, 219, 154, 187, 205, 131, 172,
        191, 148, 215, 179, 160, 202, 136, 184, 194, 168,
        217, 157, 181, 209, 144, 173, 196, 162, 218, 188,
        151, 203, 177, 139, 192, 211, 166, 183, 199, 155,
        214, 170, 186, 207, 146, 175, 197, 159, 213, 180]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, bins in zip(axes, [5, 10, 15]):
    ax.hist(runs, bins=bins, edgecolor="black")
    ax.set_title(f"{bins} bins")
    ax.set_xlabel("Runs scored")
    ax.set_ylabel("Number of matches")

fig.suptitle("IPL Match Scores: Effect of Bin Count")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
steps = rng.integers(3500, 15001, size=30)
bins = np.arange(3000, 16001, 2000)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(steps, bins=bins, edgecolor="black")
axes[0].set_title("Daily Steps — Frequency")
axes[0].set_ylabel("Number of days")

axes[1].hist(steps, bins=bins, density=True, edgecolor="black")
axes[1].set_title("Daily Steps — Density")
axes[1].set_ylabel("Probability density")

for ax in axes:
    ax.set_xlabel("Steps per day")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(4)
durations = rng.integers(120, 301, size=50)

plt.hist(durations, bins=10, color="purple", edgecolor="black")
plt.title("Spotify Song Duration Distribution")
plt.xlabel("Song duration (seconds)")
plt.ylabel("Number of songs")
plt.show()
```

```python
import matplotlib.pyplot as plt

prices = [120, 180, 250, 320, 450, 510, 650, 780, 920, 999,
          1050, 1200, 1450, 1700, 1950, 2100, 2500, 3200, 4100, 4800]

edges = [100, 501, 1001, 2001, 5001]
counts, _, _ = plt.hist(prices, bins=edges, edgecolor="black")

plt.title("Flipkart Product Prices by Range")
plt.xlabel("Price")
plt.ylabel("Number of products")
plt.xticks([100, 500, 1000, 2000, 5000])
plt.show()

for label, count in zip(
    ["100–500", "501–1000", "1001–2000", "2001–5000"], counts
):
    print(f"{label}: {int(count)} products")
```
