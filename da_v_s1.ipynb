```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [7200, 8400, 6900, 10300, 9100, 11800, 9700]  # sample counts

plt.figure(figsize=(8, 4))
plt.plot(days, steps, marker="o")
plt.title("Daily Steps Walked Over Seven Days")
plt.xlabel("Day")
plt.ylabel("Steps")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

posts = list(range(1, 11))
likes = [42, 58, 51, 75, 63, 89, 72, 96, 84, 110]  # made-up counts

plt.figure(figsize=(8, 4))
plt.plot(posts, likes, marker="o")
plt.title("Likes on the Last 10 Instagram Posts")
plt.xlabel("Post number")
plt.ylabel("Likes")
plt.xticks(posts)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
spotify_hours = [2.5, 3.0, 1.8, 4.2, 3.6, 5.1, 4.4]
youtube_hours = [1.0, 1.3, 0.9, 1.8, 1.6, 2.3, 1.4]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(days, spotify_hours, marker="o")
axes[0].set_title("Spotify Listening")
axes[0].set_xlabel("Day")
axes[0].set_ylabel("Hours")
axes[0].grid(True)

axes[1].plot(days, youtube_hours, marker="o")
axes[1].set_title("YouTube Viewing")
axes[1].set_xlabel("Day")
axes[1].set_ylabel("Hours")
axes[1].grid(True)

fig.suptitle("Daily Media Use Over One Week")
fig.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
spotify_hours = [2.5, 3.0, 1.8, 4.2, 3.6, 5.1, 4.4]
youtube_hours = [1.0, 1.3, 0.9, 1.8, 1.6, 2.3, 1.4]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(days, spotify_hours, color="#1DB954", marker="o", linewidth=2.5)
axes[0].set_title("Spotify Listening")
axes[0].set_xlabel("Day of week")
axes[0].set_ylabel("Hours listened")
axes[0].grid(True, linestyle="--", alpha=0.4)

axes[1].plot(days, youtube_hours, color="#D62828", marker="o", linewidth=2.5)
axes[1].set_title("YouTube Viewing")
axes[1].set_xlabel("Day of week")
axes[1].set_ylabel("Hours viewed")
axes[1].grid(True, linestyle="--", alpha=0.4)

fig.suptitle("Weekly Media Use", fontsize=15, fontweight="bold")
fig.tight_layout()
plt.show()
```
