```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
users = [1200, 1350, 1420, 1380, 1590, 1710,
         1840, 1790, 2010, 2250, 2130, 2180]

peak = users.index(max(users))

plt.figure(figsize=(10, 5))
plt.plot(months, users, marker="o", color="#1DB954")
plt.annotate(
    "Highest users",
    xy=(peak, users[peak]),
    xytext=(peak - 3, users[peak] + 180),
    arrowprops={"arrowstyle": "->", "color": "black"},
    fontsize=10
)
plt.title("Monthly Active Music App Users")
plt.xlabel("Month")
plt.ylabel("Active users")
plt.ylim(1000, 2600)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Apr", "May", "Jun", "Jul", "Aug", "Sep"]
orders = [8, 10, 9, 5, 12, 11]
lowest = orders.index(min(orders))

plt.figure(figsize=(9, 5))
plt.bar(months, orders, color="#E23744")
plt.annotate(
    "Lowest orders",
    xy=(lowest, orders[lowest]),
    xytext=(lowest + 0.5, 11),
    arrowprops={"arrowstyle": "->", "color": "black"},
    fontsize=10
)
plt.title("Monthly Food Delivery Orders")
plt.xlabel("Month")
plt.ylabel("Number of orders")
plt.ylim(0, 14)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

methods = ["UPI", "Card", "Wallet", "Cash"]
payments = [48, 27, 15, 10]

fig, ax = plt.subplots(figsize=(9, 6))
wedges, _, _ = ax.pie(
    payments, autopct="%1.1f%%", startangle=90
)
ax.set_title("E-commerce Payment Methods")
ax.legend(
    wedges, methods, title="Payment method",
    loc="lower right", bbox_to_anchor=(1.25, 0)
)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

movies = ["Film A", "Film B", "Film C", "Film D",
          "Film E", "Film F", "Film G", "Film H"]
ratings = [3.8, 4.2, 4.0, 4.6, 3.5, 4.9, 4.3, 4.5]
reviews = [850, 1200, 980, 1750, 620, 2300, 1450, 1620]

highest = ratings.index(max(ratings))

plt.figure(figsize=(9, 5))
plt.scatter(ratings, reviews, color="teal", s=85)
plt.annotate(
    f"Highest rating: {movies[highest]} ({ratings[highest]})",
    xy=(ratings[highest], reviews[highest]),
    xytext=(3.55, 2050),
    arrowprops={"arrowstyle": "->"},
    bbox={"boxstyle": "round,pad=0.4", "facecolor": "lightyellow",
          "edgecolor": "gray"}
)
plt.title("Movie Ratings vs Number of Reviews")
plt.xlabel("Rating")
plt.ylabel("Number of reviews")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
