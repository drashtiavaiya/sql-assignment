```python
import pandas as pd

january_orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "restaurant": ["Spice Villa", "Pizza Corner", "Green Leaf"],
    "amount": [450, 700, 320],
    "month": ["January"] * 3
})

february_orders = pd.DataFrame({
    "order_id": [104, 105, 106],
    "restaurant": ["Cafe Aroma", "Spice Villa", "Burger House"],
    "amount": [280, 550, 400],
    "month": ["February"] * 3
})

all_orders = pd.concat(
    [january_orders, february_orders],
    ignore_index=True
)

print(all_orders)
```

```python
product_details = pd.DataFrame({
    "product_id": [101, 102, 103, 104],
    "item": ["Wireless Mouse", "Laptop Stand", "Power Bank", "Keyboard"],
    "price": [599, 899, 1499, 799]
})

product_ratings = pd.DataFrame({
    "rating": [4.5, 4.2, 4.7, 4.0]
})

products_with_ratings = pd.concat(
    [product_details, product_ratings],
    axis=1
)

print(products_with_ratings)
```

```python
users = ["Aarav", "Diya", "Kabir"]
days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

daily_streams = pd.DataFrame([
    {"user": user, "day": day, "streams": 100 + user_num * 50 + day_num * 10}
    for user_num, user in enumerate(users)
    for day_num, day in enumerate(days)
])

streams_by_user = daily_streams.pivot(
    index="user",
    columns="day",
    values="streams"
)

print(streams_by_user)
```

```python
zomato_orders = pd.DataFrame({
    "restaurant": [
        "Spice Villa", "Spice Villa", "Pizza Corner", "Pizza Corner",
        "Spice Villa", "Spice Villa", "Pizza Corner", "Pizza Corner"
    ],
    "month": [
        "January", "January", "January", "January",
        "February", "February", "February", "February"
    ],
    "total_amount": [400, 600, 300, 500, 550, 750, 450, 650]
})

average_orders = zomato_orders.pivot_table(
    index="restaurant",
    columns="month",
    values="total_amount",
    aggfunc="mean"
)

print(average_orders)
```

```python
monthly_sales = {
    "sales_jan.csv": [("T-shirt", 12000), ("Jeans", 18000), ("Shoes", 15000)],
    "sales_feb.csv": [("T-shirt", 14000), ("Jeans", 16000), ("Shoes", 17000)],
    "sales_mar.csv": [("T-shirt", 13000), ("Jeans", 20000), ("Shoes", 19000)]
}

for filename, rows in monthly_sales.items():
    pd.DataFrame(rows, columns=["product", "sales"]).to_csv(
        filename, index=False
    )

files = ["sales_jan.csv", "sales_feb.csv", "sales_mar.csv"]
combined_sales = pd.concat(
    [pd.read_csv(filename) for filename in files],
    ignore_index=True
)

total_by_product = combined_sales.groupby("product")["sales"].sum()

print(total_by_product)
```
