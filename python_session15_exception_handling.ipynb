```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero."


print(safe_divide(20, 4))
print(safe_divide(20, 0))
```

```python
try:
    number_of_reviews = int(input("Enter number of reviews: "))
    total_stars = float(input("Enter total stars: "))

    average_rating = total_stars / number_of_reviews

    print("Average rating:", average_rating)

except ValueError:
    print("Invalid input. Please enter numeric values.")
```

```python
class InvalidDurationError(Exception):
    pass


def get_playlist_duration(songs):
    total_duration = 0

    for duration in songs:
        if duration < 0:
            raise InvalidDurationError(
                "Song duration cannot be negative."
            )

        total_duration += duration

    return total_duration


song_durations = [210, 185, -30, 240]

try:
    print(
        "Total playlist duration:",
        get_playlist_duration(song_durations)
    )
except InvalidDurationError as error:
    print("Error:", error)
```

```python
try:
    item_price = float(input("Enter item price: ₹"))
    quantity = int(input("Enter quantity: "))

except ValueError:
    print("Invalid input. Enter numbers only.")

else:
    total_price = item_price * quantity
    print("Total order price: ₹", total_price)

finally:
    print("Thank you for shopping!")
```

```python
try:
    first_number = float(input("Enter first number: "))
    second_number = float(input("Enter second number: "))

    result = first_number / second_number
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")

except ValueError:
    print("Invalid input. Please enter numbers only.")

print(
    "Learning: Different exceptions can be handled separately "
    "to prevent a program from crashing."
)
```
