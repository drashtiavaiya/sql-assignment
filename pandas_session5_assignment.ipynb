```python
import pandas as pd
import numpy as np

ipl_players_data = pd.DataFrame({
    "player_name": [
        "Virat Kohli", "Rohit Sharma", "MS Dhoni",
        "Shubman Gill", "Ravindra Jadeja", "Hardik Pandya"
    ],
    "team": ["RCB", "MI", "CSK", "GT", "CSK", "MI"],
    "age": [37, 39, 45, np.nan, 37, np.nan]
})

ipl_players_data.to_csv("ipl_players.csv", index=False)

ipl_players = pd.read_csv("ipl_players.csv")

players_with_missing_age = ipl_players[
    ipl_players["age"].isnull()
]

print(players_with_missing_age)
```

```python
food_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106],
    "customer_name": [
        "Aarav", "Diya", "Kabir",
        "Meera", "Rohan", "Anaya"
    ],
    "delivery_rating": [4.5, np.nan, 3.8, 4.7, np.nan, 4.2]
})

cleaned_orders = food_orders.dropna(
    subset=["delivery_rating"]
)

print(cleaned_orders)
```

```python
flipkart_reviews = pd.DataFrame({
    "product_name": [
        "Wireless Mouse", "Bluetooth Speaker",
        "Laptop Stand", "Power Bank",
        "Smart Watch", "USB Keyboard"
    ],
    "rating": [4.2, np.nan, 3.9, 4.5, np.nan, 4.1]
})

mean_rating = flipkart_reviews["rating"].mean()

flipkart_reviews["rating"] = (
    flipkart_reviews["rating"].fillna(mean_rating)
)

print("Mean rating:", round(mean_rating, 2))
print(flipkart_reviews)
```

```python
spotify_data = pd.DataFrame({
    "song": [
        "luther", "Good Days", "Pink + White",
        "As It Was", "HUMBLE.", "Snooze"
    ],
    "artist": [
        "Kendrick Lamar & SZA", "SZA", "Frank Ocean",
        "Harry Styles", "Kendrick Lamar", "SZA"
    ],
    "duration": [177, np.nan, 184, 167, np.nan, 201]
})

spotify_data.to_csv("spotify_playlist.csv", index=False)

spotify_playlist = pd.read_csv("spotify_playlist.csv")

median_duration = spotify_playlist["duration"].median()

spotify_playlist["duration"] = (
    spotify_playlist["duration"].fillna(median_duration)
)

spotify_playlist.to_csv(
    "cleaned_spotify_playlist.csv",
    index=False
)

print("Median duration:", median_duration)
print(spotify_playlist)
```

```python
instagram_data = pd.DataFrame({
    "username": [
        "aarav_fit", "diya_art", "kabir_music",
        "meera_food", "rohan_travel", "anaya_style"
    ],
    "followers": [1200, np.nan, 5400, 3200, np.nan, 7800]
})

instagram_data.to_csv("instagram_users.csv", index=False)

instagram_users = pd.read_csv("instagram_users.csv")

instagram_users["followers"] = (
    instagram_users["followers"]
    .fillna(0)
    .astype(int)
)

instagram_users.to_csv(
    "cleaned_instagram_users.csv",
    index=False
)

print(instagram_users)
```
