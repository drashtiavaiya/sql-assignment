```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
zomato = [12, 15, 14, 18, 20, 24, 22]
swiggy = [10, 13, 12, 16, 17, 21, 19]

plt.figure(figsize=(10, 5))
plt.bar(days, zomato, color="#E23744", label="Zomato")
plt.bar(days, swiggy, bottom=zomato, color="#FC8019", label="Swiggy")
plt.title("Daily Food Orders by Platform")
plt.xlabel("Day")
plt.ylabel("Number of orders")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
spotify = [1200, 1350, 1280, 1490, 1580, 1740, 1690]
youtube = [1600, 1550, 1720, 1810, 1770, 1950, 2020]

plt.figure(figsize=(10, 5))
plt.fill_between(days, spotify, color="#1DB954", alpha=0.5, label="Spotify")
plt.fill_between(days, youtube, color="#FF3333", alpha=0.4, label="YouTube")
plt.plot(days, spotify, color="#137E3A")
plt.plot(days, youtube, color="#C52222")
plt.title("Daily Active Users by Platform")
plt.xlabel("Day")
plt.ylabel("Active users")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = list(range(1, 11))
average_minutes = [31, 29, 34, 32, 30, 36, 33, 28, 35, 31]
std_dev = [4, 3, 5, 4, 3, 6, 4, 3, 5, 4]

plt.figure(figsize=(10, 5))
plt.errorbar(days, average_minutes, yerr=std_dev,
             fmt="o-", capsize=5, color="teal",
             ecolor="gray", label="Average ± standard deviation")
plt.title("Flipkart Delivery Times Over 10 Days")
plt.xlabel("Day")
plt.ylabel("Delivery time (minutes)")
plt.xticks(days)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

dates = [datetime(2026, 9, 14) + timedelta(days=i) for i in range(14)]
balance = [850, 920, 710, 1050, 980, 1240, 1120,
           1390, 1280, 1510, 1330, 1690, 1580, 1820]

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(dates, balance, marker="o", color="#2366B8")
ax.set_title("Daily Paytm Wallet Balance")
ax.set_xlabel("Date")
ax.set_ylabel("Balance (₹)")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))
fig.autofmt_xdate()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
spotify = [1200, 1350, 1280, 1490, 1580, 1740, 1690]
youtube = [1600, 1550, 1720, 1810, 1770, 1950, 2020]

fig, ax = plt.subplots(figsize=(11, 5))
ax.fill_between(days, spotify, color="#1DB954", alpha=0.35)
ax.fill_between(days, youtube, color="#FF3333", alpha=0.25)
ax.plot(days, spotify, color="#137E3A", linewidth=2, label="Spotify")
ax.plot(days, youtube, color="#C52222", linewidth=2, label="YouTube")

ax.set_title("Daily Active Users by Platform", fontsize=14, weight="bold")
ax.set_xlabel("Day of the week")
ax.set_ylabel("Active users")
ax.set_ylim(bottom=0)
ax.legend(loc="upper left")
ax.grid(axis="y", alpha=0.25)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
