```python
add_gst = lambda price: price + (price * 0.18)

prices = [100, 250, 500]

for price in prices:
    print("Original price:", price, "| Price with GST:", add_gst(price))
```

```python
song_titles = [
    "  shape OF you ",
    "GOOD days  ",
    "  pink + WHITE",
    "as IT was "
]

cleaned_titles = list(
    map(lambda title: title.strip().title(), song_titles)
)

print(cleaned_titles)
```

```python
product_names = [
    "Samsung Mobile",
    "iPhone",
    "Sony TV",
    "speaker",
    "Laptop",
    "Smartwatch"
]

s_products = list(
    filter(lambda product: product.lower().startswith("s"), product_names)
)

print(s_products)
```

```python
from functools import reduce

order_amounts = [120, 340, 560, 80]

total_bill = reduce(
    lambda total, amount: total + amount,
    order_amounts
)

print("Total bill amount:", total_bill)
```

```python
from functools import reduce

numbers = [40, 60, 80, 120]

doubled_numbers = map(lambda number: number * 2, numbers)

filtered_numbers = filter(
    lambda number: number > 100,
    doubled_numbers
)

final_sum = reduce(
    lambda total, number: total + number,
    filtered_numbers
)

print("Final sum:", final_sum)
```
