```python
playlist_prices = {
    "Chill Hits": 199,
    "Workout Mix": 249,
    "Indie Vibes": 179,
    "Bollywood Beats": 299,
    "Late Night R&B": 229
}

print("Playlist Prices:", playlist_prices)
```

```python
def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price

update_playlist_price("Indie Vibes", 219)

print("Updated Playlist Prices:", playlist_prices)
```

```python
del playlist_prices["Workout Mix"]

print("Dictionary After Deletion:", playlist_prices)
```

```python
zomato_restaurants = {
    "Pizza Hut",
    "Burger King",
    "Subway",
    "McDonald's"
}

swiggy_restaurants = {
    "Burger King",
    "Domino's",
    "Subway",
    "KFC"
}

all_restaurants = zomato_restaurants.union(swiggy_restaurants)
common_restaurants = zomato_restaurants.intersection(swiggy_restaurants)

print("Union:", all_restaurants)
print("Intersection:", common_restaurants)
```
