```python
import pandas as pd

orders = pd.DataFrame([
    {"user": "A", "item": "Pizza", "price": 300},
    {"user": "B", "item": "Burger", "price": 150},
    {"user": "A", "item": "Burger", "price": 150},
])

print("Raw orders")
display(orders)

total_spend = (
    orders.groupby("user", as_index=False)["price"]
    .sum()
    .rename(columns={"price": "total_spend"})
)

print("Total spend per user")
display(total_spend)
```
