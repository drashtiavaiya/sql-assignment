```python
order_amounts = [120, 250, 90, 310, 150]

total_order_value = 0

for amount in order_amounts:
    total_order_value += amount

print("Total order value:", total_order_value)
```

```python
cricket_scores = [45, 78, 102, 34, 67, 89]

index = 0

while index < len(cricket_scores):
    if cricket_scores[index] > 100:
        break

    print(cricket_scores[index])
    index += 1
```

```python
item_prices = [299, 499, 199, 999, 149]

total_price = 0

for price in item_prices:
    if price < 200:
        continue

    total_price += price

print("Total of remaining items:", total_price)
```

```python
favorite_songs = [
    "Good Days",
    "Pink + White",
    "Luther",
    "Not Like Us",
    "As It Was"
]

for position, song in enumerate(favorite_songs, start=1):
    print(position, "-", song)
```

```python
follower_counts = [120, 1500, 23000, 800, 45000]

for followers in follower_counts:
    if followers < 1000:
        category = "Micro"
    elif followers <= 10000:
        category = "Influencer"
    else:
        category = "Celebrity"

    print(followers, "-", category)
```
