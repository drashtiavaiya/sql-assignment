```python
songs = [
    "Good Days",
    "Pink + White",
    "Luther",
    "Not Like Us",
    "As It Was"
]

with open("my_playlist.txt", "w") as file:
    for song in songs:
        file.write(song + "\n")

print("my_playlist.txt created successfully.")
```

```python
with open("my_playlist.txt", "r") as file:
    for song in file:
        print(song.strip().upper())
```

```python
import csv

matches = [
    ["Match", "Team1", "Team2", "Winner"],
    [1, "Mumbai Indians", "Chennai Super Kings", "Mumbai Indians"],
    [2, "Royal Challengers Bengaluru", "Rajasthan Royals", "Rajasthan Royals"],
    [3, "Delhi Capitals", "Gujarat Titans", "Gujarat Titans"]
]

with open("ipl_scores.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(matches)

with open("ipl_scores.csv", "r") as file:
    reader = csv.DictReader(file)

    for match in reader:
        print("Match", match["Match"], "Winner:", match["Winner"])
```

```python
import json

profile = {
    "username": "drashti",
    "followers": 12500,
    "bio": "Learning Python and data analytics"
}

with open("user_profile.json", "w") as file:
    json.dump(profile, file, indent=4)

with open("user_profile.json", "r") as file:
    user_data = json.load(file)

print("Username:", user_data["username"])
print("Followers:", user_data["followers"])
```
