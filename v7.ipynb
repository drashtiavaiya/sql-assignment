```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [7200, 8100, 6500, 9800, 8400, 11100, 8700]
water = [1.8, 2.1, 1.6, 2.4, 2.0, 2.7, 2.2]

plt.figure(figsize=(9, 7))

plt.subplot(2, 1, 1)
plt.plot(days, steps, marker="o")
plt.title("Daily Steps")
plt.ylabel("Steps")
plt.grid(True, alpha=0.3)

plt.subplot(2, 1, 2)
plt.plot(days, water, marker="o", color="teal")
plt.title("Daily Water Intake")
plt.xlabel("Day")
plt.ylabel("Litres")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [12, 18, 15, 21]

plt.figure(figsize=(11, 8))

plt.subplot(2, 2, 1)
plt.plot(quarters, posts, marker="o")
plt.title("Line Chart")
plt.ylabel("Posts")

plt.subplot(2, 2, 2)
plt.bar(quarters, posts, color="coral")
plt.title("Bar Chart")
plt.ylabel("Posts")

plt.subplot(2, 2, 3)
plt.scatter(quarters, posts, color="purple", s=90)
plt.title("Scatter Plot")
plt.xlabel("Quarter")
plt.ylabel("Posts")

plt.subplot(2, 2, 4)
plt.pie(posts, labels=quarters, autopct="%1.1f%%", startangle=90)
plt.title("Pie Chart")

plt.suptitle("Instagram Posts by Quarter")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Apr", "May", "Jun", "Jul", "Aug", "Sep"]
orders = {
    "Zomato": [8, 10, 9, 12, 11, 14],
    "Swiggy": [6, 7, 8, 9, 8, 10],
    "Domino's": [2, 3, 2, 4, 3, 5],
}

fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)

for ax, (app, counts) in zip(axes, orders.items()):
    ax.plot(months, counts, marker="o")
    ax.set_title(app)
    ax.set_xlabel("Month")
    ax.grid(True, alpha=0.3)

axes[0].set_ylabel("Number of orders")
fig.suptitle("Monthly Food Orders by App")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Apr", "May", "Jun", "Jul", "Aug", "Sep"]
usage_gb = {
    "YouTube": [8, 9, 11, 10, 13, 12],
    "Instagram": [5, 6, 7, 8, 7, 9],
    "Spotify": [2, 2.5, 3, 2.8, 3.5, 4],
    "WhatsApp": [1, 1.2, 1.4, 1.3, 1.7, 1.8],
}

fig, axes = plt.subplots(2, 2, figsize=(11, 7), sharex=True, sharey=True)

for ax, (app, usage) in zip(axes.flat, usage_gb.items()):
    ax.plot(months, usage, marker="o")
    ax.set_title(app)
    ax.grid(True, alpha=0.3)

for ax in axes[:, 0]:
    ax.set_ylabel("Data used (GB)")
for ax in axes[1, :]:
    ax.set_xlabel("Month")

fig.suptitle("Monthly Mobile Data Usage")
plt.tight_layout()
plt.show()
```
