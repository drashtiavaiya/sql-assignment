```python
food_items = [
    "Pizza",
    "Burger",
    "Pasta",
    "Biryani"
]

prices = [250, 180, 220, 300]

for food, price in zip(food_items, prices):
    print(food, "- ₹", price)
```

```python
def instagram_posts_generator(posts):
    for caption in posts:
        yield caption


post_captions = [
    "Morning coffee and calm vibes",
    "Learning Python one step at a time",
    "Weekend memories",
    "Sunset views"
]

post_generator = instagram_posts_generator(post_captions)

while True:
    try:
        print("Next post:", next(post_generator))
    except StopIteration:
        print("All captions have been displayed.")
        break
```

```python
def cashback_generator(transactions):
    for amount in transactions:
        yield amount * 0.05


paytm_transactions = [500, 1200, 750, 2000]

for cashback in cashback_generator(paytm_transactions):
    print("Cashback earned: ₹", cashback)
```
