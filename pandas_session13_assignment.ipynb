```python
import pandas as pd

url = "https://raw.githubusercontent.com/amar08007/ipl-career-dataset/main/data/ipl_batting_career_progression.csv"

season_data = pd.read_csv(url)
batsmen = season_data.loc[
    season_data["season"] == 2023,
    ["player_name", "innings", "runs_scored",
     "batting_average", "batting_strike_rate"]
].rename(columns={
    "player_name": "Player",
    "innings": "Innings",
    "runs_scored": "Runs",
    "batting_average": "Average",
    "batting_strike_rate": "Strike Rate"
})

batsmen.to_csv("IPL_2023_Batsmen.csv", index=False)
print(batsmen.head(10))
```

```python
runs_column = next(
    column for column in batsmen.columns
    if column.strip().lower() == "runs"
)

runs = pd.to_numeric(
    batsmen[runs_column].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
).dropna()

print("Mean:", runs.mean())
print("Median:", runs.median())
print("Mode:", runs.mode().tolist())
print("Minimum:", runs.min())
print("Maximum:", runs.max())
```

```python
restaurants = pd.DataFrame({
    "name": ["Spice Villa", "Pizza Corner", "Green Leaf",
             "Cafe Aroma", "Royal Kitchen"],
    "rating": [4.5, 4.0, 4.7, 3.8, 4.3],
    "votes": [850, 420, 1200, 300, 680],
    "avg_cost": [700, 450, 900, 350, 650]
})

rating_votes_correlation = restaurants["rating"].corr(
    restaurants["votes"]
)

print(restaurants)
print("\nCorrelation between rating and votes:",
      round(rating_votes_correlation, 3))
```

```python
print("IPL batsmen: numerical summary")
display(batsmen.describe())

median_runs = batsmen["Runs"].median()
highest_runs = batsmen["Runs"].max()
print(f"Insight: The highest runs ({highest_runs:.0f}) are "
      f"{highest_runs - median_runs:.0f} above the median ({median_runs:.0f}).")
```

```python
print("Two additional analyses:")
print("1. Univariate: examine the distribution of batting strike rates.")
print("2. Bivariate: examine the relationship between innings and runs.")

print("\nAnalysis 2 — innings and runs:")
display(batsmen[["Innings", "Runs"]].corr())

correlation = batsmen["Innings"].corr(batsmen["Runs"])
print(f"Correlation: {correlation:.2f}")
```
