```python
import matplotlib.pyplot as plt

friends = ["Aarav", "Diya", "Kabir", "Mira", "Rohan", "Anaya",
           "Ishaan", "Tara", "Neel", "Sara", "Dev", "Riya"]
orders = [2, 5, 4, 8, 3, 10, 6, 7, 12, 9, 5, 11]
spent = [380, 920, 700, 1450, 540, 1820,
         1080, 1300, 2250, 1640, 850, 1980]

plt.figure(figsize=(10, 6))
plt.scatter(orders, spent, color="teal")

for name, x, y in zip(friends, orders, spent):
    plt.annotate(name, (x, y), xytext=(5, 5),
                 textcoords="offset points", fontsize=8)

plt.title("Zomato Orders vs Amount Spent")
plt.xlabel("Number of orders")
plt.ylabel("Amount spent (₹)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

orders = np.array([2, 5, 4, 8, 3, 10, 6, 7, 12, 9, 5, 11])
spent = np.array([380, 920, 700, 1450, 540, 1820,
                  1080, 1300, 2250, 1640, 850, 1980])

slope, intercept = np.polyfit(orders, spent, 1)
x_line = np.linspace(orders.min(), orders.max(), 100)
y_line = slope * x_line + intercept

plt.scatter(orders, spent, label="Friends")
plt.plot(x_line, y_line, color="red",
         label=f"Trend: y = {slope:.1f}x + {intercept:.1f}")
plt.title("Zomato Orders vs Amount Spent")
plt.xlabel("Number of orders")
plt.ylabel("Amount spent (₹)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"Trend line: amount spent = {slope:.2f} × orders + {intercept:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

youtube_hours = np.array([1.0, 2.5, 3.0, 4.5, 5.0, 6.0, 7.5, 8.0,
                          9.5, 10.0, 11.5, 12.0, 13.5, 14.0, 15.0])
spotify_hours = np.array([2.0, 1.5, 3.5, 3.0, 5.0, 4.5, 6.0, 5.5,
                          7.5, 7.0, 8.5, 9.0, 8.0, 10.5, 9.5])

slope, intercept = np.polyfit(youtube_hours, spotify_hours, 1)
x_line = np.linspace(youtube_hours.min(), youtube_hours.max(), 100)

plt.scatter(youtube_hours, spotify_hours, color="teal", label="Users")
plt.plot(x_line, slope * x_line + intercept,
         color="red", label="Regression line")
plt.title("YouTube vs Spotify Listening Hours")
plt.xlabel("YouTube hours")
plt.ylabel("Spotify hours")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"Regression line: Spotify hours = {slope:.2f} × YouTube hours + {intercept:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

orders = np.array([1, 3, 2, 5, 4, 7, 6, 9, 8, 10])
total_spent = np.array([450, 1250, 820, 2100, 1700,
                        2900, 2450, 3700, 3250, 4100])

slope, intercept = np.polyfit(orders, total_spent, 1)
x_line = np.linspace(orders.min(), orders.max(), 100)

plt.scatter(orders, total_spent, color="purple", label="Users")
plt.plot(x_line, slope * x_line + intercept,
         color="orange", label="Regression line")
plt.title("Flipkart Orders vs Total Amount Spent")
plt.xlabel("Number of orders")
plt.ylabel("Total amount spent (₹)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"Trend line: total spent = {slope:.2f} × orders + {intercept:.2f}")
```
