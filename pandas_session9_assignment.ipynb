```python
import pandas as pd

dates = pd.Series([
    "05-01-2026",
    "18-03-2025",
    "27-07-2024",
    "09-11-2023",
    "22-09-2026"
])

converted_dates = pd.to_datetime(dates, format="%d-%m-%Y")

print(converted_dates)
print("\nData type:", converted_dates.dtype)
```

```python
movies = pd.DataFrame({
    "movie": ["Dangal", "3 Idiots", "Gully Boy", "Queen", "Andhadhun"],
    "release_date": [
        "2016-12-23", "2009-12-25", "2019-02-14",
        "2014-03-07", "2018-10-05"
    ]
})

movies["release_date"] = pd.to_datetime(movies["release_date"])

movies["year"] = movies["release_date"].dt.year
movies["month"] = movies["release_date"].dt.month
movies["day"] = movies["release_date"].dt.day

print(movies)
```

```python
today = pd.Timestamp.now(tz="Asia/Kolkata").normalize()

next_seven_days = pd.date_range(
    start=today,
    periods=7,
    freq="D"
)

print(next_seven_days)
```

```python
subscriptions = pd.DataFrame({
    "user_id": [101, 102, 103, 104],
    "start_date": ["2024-01-15", "2025-06-01", "2026-02-10", "2026-08-20"]
})

subscriptions["start_date"] = pd.to_datetime(subscriptions["start_date"])

today_date = pd.Timestamp.now(
    tz="Asia/Kolkata"
).tz_localize(None).normalize()

subscriptions["tenure_days"] = (
    today_date - subscriptions["start_date"]
).dt.days

print(subscriptions)
```
