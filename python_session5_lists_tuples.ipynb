```python
def remove_last_item(order_list):
    removed_item = order_list.pop()
    return removed_item

sample_order = ["Pizza", "Burger", "Cold Drink"]

removed = remove_last_item(sample_order)

print("Removed Item:", removed)
print("Updated Order:", sample_order)
```
