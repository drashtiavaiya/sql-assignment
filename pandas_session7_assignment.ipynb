```python
import pandas as pd

sales_data = pd.DataFrame({
    "ProductName": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "Price": [45000, 600, 1200, 12000, 1800],
    "Qty": [2, 5, 3, 1, 4]
})

sales_data.to_csv("flipkart_product_sales.csv", index=False)

sales = pd.read_csv("flipkart_product_sales.csv")
sales["TotalValue"] = sales["Price"] * sales["Qty"]

print(sales)
```

```python
food_items = pd.DataFrame({
    "Item": ["Pizza", "Burger", "Pasta", "Sandwich", "Biryani"],
    "Price": [400, 200, 350, 150, 300],
    "Qty": [2, 3, 1, 4, 2]
})

food_items["DiscountedPrice"] = food_items["Price"].apply(
    lambda price: price * 0.90
)

print(food_items)
```

```python
instagram_posts = pd.DataFrame({
    "PostID": [101, 102, 103, 104],
    "Likes": [1500, 2000000, 850, 12500],
    "Comments": [250, 3200, 1500000, 75]
})

def format_count(number):
    if number >= 1000000:
        return f"{number / 1000000:g}M"
    if number >= 1000:
        return f"{number / 1000:g}K"
    return str(number)

instagram_posts[["Likes", "Comments"]] = (
    instagram_posts[["Likes", "Comments"]].applymap(format_count)
)

print(instagram_posts)
```

```python
zomato_orders = pd.DataFrame({
    "Restaurant": ["Pizza Corner", "Green Leaf", "Spice Villa", "Cafe Aroma"],
    "Price": [400, 250, 350, 180],
    "Qty": [2, 3, 1, 4],
    "DeliveryCharge": [40, 30, 35, 25]
})

zomato_orders["FinalAmount"] = zomato_orders.apply(
    lambda row: row["Price"] * row["Qty"] + row["DeliveryCharge"],
    axis=1
)

print(zomato_orders)
```

```python
myntra_orders = pd.DataFrame({
    "Item": ["T-shirt", "Jeans", "Jacket", "Shoes"],
    "Price": [800, 1500, 2500, 2000],
    "Qty": [2, 1, 1, 2]
})

myntra_orders["GST"] = myntra_orders["Price"] * myntra_orders["Qty"] * 0.05

print(myntra_orders)

# I learned that multiplying Price by Qty gives the order value.
# Applying 0.05 to that value calculates 5% GST for each order.
```
