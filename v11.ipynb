```python
import matplotlib.pyplot as plt
from google.colab import files

restaurants = ["Spice Garden", "Green Bowl", "Coastal Kitchen",
               "Urban Pizza", "Chai Corner"]
orders = [42, 35, 28, 51, 39]

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(restaurants, orders, color="teal")
ax.set_title("Zomato Orders by Restaurant")
ax.set_ylabel("Number of orders")
ax.tick_params(axis="x", rotation=20)
fig.tight_layout()

fig.savefig("v11b.png", dpi=150)
plt.show()
files.download("v11b.png")
```

```python
import matplotlib.pyplot as plt
from google.colab import files

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [7200, 8100, 6500, 9800, 8400, 11100, 8700]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(days, steps, marker="o", color="green")
ax.set_title("Daily Steps Over One Week")
ax.set_xlabel("Day")
ax.set_ylabel("Steps")
ax.grid(True, alpha=0.3)
fig.tight_layout()

fig.savefig("v11l.svg")
plt.show()
files.download("v11l.svg")
```

```python
import matplotlib.pyplot as plt
from google.colab import files

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
time_share = [30, 28, 24, 18]

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(time_share, labels=apps, autopct="%1.1f%%", startangle=90)
ax.set_title("Daily App Time Share")
fig.tight_layout()

fig.savefig("v11p.pdf")
plt.show()
files.download("v11p.pdf")
```
