```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
active_users = [1200, 1350, 1280, 1490, 1580, 1740, 1690]

plt.figure(figsize=(10, 4))
plt.plot(days, active_users, marker="o", linewidth=2)
plt.title("Daily Active Music App Users")
plt.xlabel("Day")
plt.ylabel("Active users")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
zomato = [12, 15, 14, 18, 20, 24, 22]
swiggy = [10, 13, 12, 16, 17, 21, 19]

plt.figure(figsize=(10, 4))
plt.plot(days, zomato, linestyle="--", marker="o", label="Zomato")
plt.plot(days, swiggy, linestyle=":", marker="s", label="Swiggy")
plt.title("Daily Food Orders")
plt.xlabel("Day")
plt.ylabel("Number of orders")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.style.use("ggplot")

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [85, 92, 88, 104, 117, 130, 125]

plt.figure(figsize=(10, 4))
plt.plot(days, sales, color="orange", marker="o", linewidth=2)
plt.title("Daily Flipkart Sales")
plt.xlabel("Day")
plt.ylabel("Sales (sample units)")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.style.use("default")

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
posts = [3, 5, 4, 6, 5, 8, 7]
colors = ["#F58529", "#DD2A7B", "#8134AF", "#515BD4",
          "#FEDA77", "#D62976", "#962FBF"]

plt.figure(figsize=(10, 4))
plt.bar(days, posts, color=colors, edgecolor="#333333")
plt.title("Daily Instagram Posts")
plt.xlabel("Day")
plt.ylabel("Posts uploaded")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.style.use("dark_background")

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
zomato = [12, 15, 14, 18, 20, 24, 22]
swiggy = [10, 13, 12, 16, 17, 21, 19]

plt.figure(figsize=(10, 4))
plt.plot(days, zomato, color="#00E5FF", linestyle="--",
         marker="o", linewidth=2, label="Zomato")
plt.plot(days, swiggy, color="#D6FF5C", linestyle=":",
         marker="s", linewidth=2, label="Swiggy")
plt.title("Daily Food Orders — Dark Theme")
plt.xlabel("Day")
plt.ylabel("Number of orders")
plt.legend()
plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()
```
