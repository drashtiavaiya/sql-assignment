```python
import pandas as pd

users = pd.DataFrame({
    "user_id": [1, 2, 3, 4, 5],
    "username": ["Aarav", "Diya", "Kabir", "Meera", "Rohan"],
    "city": ["Surat", "Mumbai", "Delhi", "Pune", "Ahmedabad"]
})

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "user_id": [1, 2, 1, 4, 2],
    "order_amount": [450, 700, 300, 550, 250]
})

users_with_orders = pd.merge(users, orders, on="user_id", how="inner")

print(users_with_orders)
```

```python
all_users_and_orders = pd.merge(
    users,
    orders,
    on="user_id",
    how="left"
)

print(all_users_and_orders)
```

```python
restaurants = pd.DataFrame({
    "restaurant_id": [1, 2, 3, 4],
    "restaurant_name": ["Spice Villa", "Pizza Corner", "Green Leaf", "Cafe Aroma"],
    "city": ["Surat", "Surat", "Ahmedabad", "Vadodara"]
})

zomato_reviews = pd.DataFrame({
    "review_id": [201, 202, 203, 204],
    "restaurant_id": [1, 2, 2, 5],
    "rating": [4.5, 4.0, 3.8, 4.2],
    "reviewer_city": ["Surat", "Mumbai", "Surat", "Pune"]
})

all_restaurants_and_reviews = pd.merge(
    restaurants,
    zomato_reviews,
    on="restaurant_id",
    how="outer"
)

print(all_restaurants_and_reviews)
```

```python
playlists = pd.DataFrame({
    "playlist_id": [101, 102, 103, 104, 105],
    "user_id": [1, 2, 3, 4, 5],
    "genre": ["Hip-hop", "Pop", "Indie", "R&B", "Rock"],
    "city": ["Surat", "Mumbai", "Jaipur", "Pune", "Ahmedabad"]
})

user_profiles = pd.DataFrame({
    "user_id": [1, 2, 3, 4, 5],
    "username": ["Aarav", "Diya", "Kabir", "Meera", "Rohan"],
    "city": ["Surat", "Mumbai", "Delhi", "Pune", "Ahmedabad"]
})

matching_playlists = pd.merge(
    playlists,
    user_profiles,
    on=["user_id", "city"],
    how="inner"
)

print(matching_playlists)
```

```python
products = pd.DataFrame({
    "product_id": [1, 2, 3],
    "name": ["Wireless Mouse", "Laptop Stand", "Power Bank"],
    "category": ["Electronics", "Accessories", "Electronics"]
})

ratings = pd.DataFrame({
    "rating_id": [101, 102, 103],
    "product_id": [1, 2, 4],
    "rating": [4.5, 4.0, 3.5]
})

default_merge = pd.merge(products, ratings, on="product_id")
print("Default merge:")
print(default_merge)

right_join = pd.merge(
    products, ratings, on="product_id", how="right"
)
print("\nRight join:")
print(right_join)
```
