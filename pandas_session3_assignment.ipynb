```python
spotify_songs = pd.DataFrame({
    "song": [
        "luther", "Good Days", "Pink + White", "As It Was",
        "HUMBLE.", "Nights", "Snooze", "See You Again"
    ],
    "artist": [
        "Kendrick Lamar & SZA", "SZA", "Frank Ocean", "Harry Styles",
        "Kendrick Lamar", "Frank Ocean", "SZA", "Tyler, The Creator"
    ],
    "streams": [
        2500000, 1800000, 950000, 3200000,
        4100000, 1250000, 2800000, 1750000
    ],
    "duration": [177, 278, 184, 167, 177, 307, 201, 180]
})

popular_short_songs = spotify_songs.query(
    "streams > 1000000 and duration < 180"
)

print(popular_short_songs)
```
