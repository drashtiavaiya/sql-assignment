```python
import pandas as pd
from sqlalchemy import create_engine, URL
from getpass import getpass

# Read the exported Mumbai Indians matches
matches = pd.read_csv("mi_matches.csv")

# Keep only the matches won by Mumbai Indians
mi_wins = matches[matches["winner"] == "Mumbai Indians"]

# Display the filtered rows
display(mi_wins)

# Connect to the MySQL database
password = getpass("Enter your MySQL password: ")

connection_url = URL.create(
    "mysql+pymysql",
    username="root",
    password=password,
    host="localhost",
    database="session20_db"
)

engine = create_engine(connection_url)

# Insert the filtered data into a new SQL table
mi_wins.to_sql(
    "mi_wins",
    con=engine,
    if_exists="replace",
    index=False
)

print(f"{len(mi_wins)} rows inserted into the mi_wins table.")
```
