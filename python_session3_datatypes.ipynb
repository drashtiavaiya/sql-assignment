```python
order_price = input("Enter the Zomato order price: ₹")

price = float(order_price)
gst = price * 0.18
final_bill = price + gst

print("Price:", price)
print("GST (18%):", round(gst, 2))
print("Final Bill:", round(final_bill, 2))
```

```python
product_prices = ["799.99", "2995.0", "150"]

prices_as_float = [float(price) for price in product_prices]
total_cart_value = sum(prices_as_float)

print("Converted Prices:", prices_as_float)
print("Total Cart Value: ₹", total_cart_value)
```

```python
def is_discount_applicable(order_amount):
    return order_amount > 500

print("Order ₹450:", is_discount_applicable(450))
print("Order ₹750:", is_discount_applicable(750))
```

```python
ratings = ["4.5", "3.0", "5", "4.2"]

ratings_as_float = [float(rating) for rating in ratings]
highest_rating = max(ratings_as_float)

print("Converted Ratings:", ratings_as_float)
print("Highest Rating:", highest_rating)
```
