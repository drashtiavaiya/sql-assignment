```python
class Product:
    def get_discount(self):
        return 0


class Electronics(Product):
    def get_discount(self):
        return 10


regular_product = Product()
electronic_product = Electronics()

print("Regular product discount:", regular_product.get_discount(), "%")
print("Electronics discount:", electronic_product.get_discount(), "%")
```

```python
class FoodOrder:
    def __init__(self, base_price):
        self.base_price = base_price

    def calculate_total(self):
        return self.base_price


class ZomatoOrder(FoodOrder):
    def calculate_total(self):
        delivery_charge = self.base_price * 0.05
        return self.base_price + delivery_charge


food_order = FoodOrder(1000)
zomato_order = ZomatoOrder(1000)

print("Food order total: ₹", food_order.calculate_total())
print("Zomato order total: ₹", zomato_order.calculate_total())
```

```python
class Influencer:
    def bonus(self):
        return 2000


class BrandManager:
    def bonus(self):
        return 5000


def show_bonus(employee):
    print("Bonus: ₹", employee.bonus())


influencer = Influencer()
brand_manager = BrandManager()

show_bonus(influencer)
show_bonus(brand_manager)
```

```python
class User:
    def get_status(self):
        return "active"


class PremiumUser(User):
    def get_status(self):
        return "premium"


regular_user = User()
premium_user = PremiumUser()

print("Regular user status:", regular_user.get_status())
print("Premium user status:", premium_user.get_status())
```
