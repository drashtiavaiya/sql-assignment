```python
import pandas as pd

movies = pd.DataFrame({
    "title": [
        "Dangal", "3 Idiots", "Gully Boy",
        "Queen", "Andhadhun", "Zindagi Na Milegi Dobara"
    ],
    "rating": [8.3, 8.4, 7.9, 8.1, 8.2, 8.2],
    "release_year": [2016, 2009, 2019, 2013, 2018, 2011]
})

sorted_movies = movies.sort_values(
    by="rating",
    ascending=False
)

print(sorted_movies)
```

```python
ipl_teams = pd.DataFrame({
    "team": ["CSK", "MI", "RCB", "KKR", "RR", "GT"],
    "points": [18, 16, 14, 12, 10, 8],
    "net_run_rate": [0.652, 0.421, 0.315, -0.125, -0.280, -0.450]
})

sorted_teams = ipl_teams.sort_index(ascending=False)

print(sorted_teams)
```

```python
youtube_videos = pd.DataFrame({
    "video_title": [
        "Python Tutorial", "Data Analytics Roadmap",
        "Pandas Basics", "SQL Interview Questions",
        "Power BI Dashboard"
    ],
    "views": [250000, 180000, 320000, 410000, 275000],
    "likes": [15000, 12000, 21000, 28000, 17500]
}, index=[10, 20, 30, 40, 50])

reset_videos = youtube_videos.reset_index(drop=True)

print(reset_videos)
```

```python
zomato_restaurants = pd.DataFrame({
    "restaurant": [
        "Pizza Corner", "Green Leaf", "Cafe Aroma",
        "Spice Villa", "Royal Kitchen", "Tasty Bites"
    ],
    "rating": [4.0, 4.5, 4.3, 4.8, 4.1, 3.9]
})

reindexed_restaurants = zomato_restaurants.reindex(
    [3, 0, 1, 2, 4, 5]
)

print(reindexed_restaurants)
```

```python
flipkart_products = pd.DataFrame({
    "product_name": [
        "Wireless Mouse", "Bluetooth Speaker", "Laptop Stand",
        "Power Bank", "Smart Watch", "USB Keyboard"
    ],
    "price": [599, 1299, 899, 1499, 2499, 799],
    "discount": [20, 35, 15, 40, 30, 25]
})

sorted_products = (
    flipkart_products
    .sort_values(by="discount", ascending=False)
    .reset_index(drop=True)
)

print(sorted_products)
```
