```python
import pandas as pd

ipl_url = "https://gist.githubusercontent.com/VishnuVelukutty/fc6d8d6731ec20cabc6614b30c31a388/raw/4fb3a59155085850a281ffe66b736c4d273e2fb3/deliveries.csv"

ipl = pd.read_csv(ipl_url)

runs_by_team = (
    ipl.groupby("batting_team")["total_runs"]
    .sum()
    .sort_values(ascending=False)
)

print(runs_by_team)
```

```python
zomato_orders = pd.DataFrame({
    "city": [
        "Surat", "Surat", "Surat", "Surat",
        "Mumbai", "Mumbai", "Mumbai", "Mumbai"
    ],
    "restaurant": [
        "Spice Villa", "Spice Villa", "Pizza Corner", "Pizza Corner",
        "Cafe Aroma", "Cafe Aroma", "Green Leaf", "Green Leaf"
    ],
    "order_amount": [400, 600, 300, 500, 550, 750, 450, 650]
})

average_order_amount = zomato_orders.groupby(
    ["city", "restaurant"]
)["order_amount"].mean()

print(average_order_amount)
```

```python
flipkart_sales = pd.DataFrame({
    "category": [
        "Electronics", "Electronics", "Fashion", "Fashion",
        "Electronics", "Electronics", "Fashion", "Fashion"
    ],
    "region": [
        "West", "West", "West", "West",
        "North", "North", "North", "North"
    ],
    "sales": [12000, 18000, 7000, 9000, 15000, 21000, 8000, 10000],
    "units_sold": [5, 7, 12, 16, 6, 8, 14, 18]
})

sales_summary = flipkart_sales.groupby(
    ["category", "region"]
).agg(
    total_sales=("sales", "sum"),
    average_units_sold=("units_sold", "mean")
)

print(sales_summary)
```

```python
regional_sales = pd.DataFrame({
    "region": ["West", "West", "West", "North", "North", "North"],
    "revenue": [12000, 18000, 9000, 15000, 21000, 11000]
})

revenue_summary = regional_sales.groupby("region").agg({
    "revenue": ["sum", "max", "min"]
})

print(revenue_summary)
```

```python
spotify_streams = pd.DataFrame({
    "artist": [
        "SZA", "SZA", "Frank Ocean", "Frank Ocean",
        "Kendrick Lamar", "Kendrick Lamar", "Harry Styles", "Harry Styles"
    ],
    "genre": [
        "R&B", "R&B", "R&B", "R&B",
        "Hip-hop", "Hip-hop", "Pop", "Pop"
    ],
    "streams": [
        1800000, 2500000, 1200000, 1600000,
        4100000, 3200000, 2800000, 2200000
    ]
})

genre_summary = (
    spotify_streams.groupby("genre")["streams"]
    .agg(total_streams="sum", average_streams="mean")
    .sort_values("total_streams", ascending=False)
)

print(genre_summary)
```
