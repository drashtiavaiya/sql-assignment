```python
import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [140, 125, 78, 52, 91]  # illustrative daily orders
colors = ["#E63946", "#F4A261", "#457B9D", "#8D5A97", "#2A9D8F"]

plt.figure(figsize=(9, 4))
plt.bar(apps, orders, color=colors)
plt.title("Daily Orders Across Five Food Apps")
plt.xlabel("Food app")
plt.ylabel("Orders")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [140, 125, 78, 52, 91]
colors = ["#E63946", "#F4A261", "#457B9D", "#8D5A97", "#2A9D8F"]

plt.figure(figsize=(9, 4))
plt.bar(apps, orders, color=colors)

for position, count in enumerate(orders):
    plt.text(position, count + 2, str(count), ha="center", va="bottom")

plt.ylim(0, max(orders) + 20)
plt.title("Daily Orders Across Five Food Apps")
plt.xlabel("Food app")
plt.ylabel("Orders")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [140, 125, 78, 52, 91]
colors = ["#E63946", "#F4A261", "#457B9D", "#8D5A97", "#2A9D8F"]

plt.figure(figsize=(9, 4))
plt.bar(apps, orders, width=0.45, color=colors)

for position, count in enumerate(orders):
    plt.text(position, count + 2, str(count), ha="center", va="bottom")

plt.ylim(0, max(orders) + 20)
plt.title("Daily Orders Across Five Food Apps")
plt.xlabel("Food app")
plt.ylabel("Orders")
plt.tight_layout()
plt.show()
```
